## Observations

By comparing genuine and fraudulent job postings, several recurring patterns become apparent.

Fraudulent postings frequently contain:

- unrealistic salary claims
- urgency-based language
- requests for payment
- vague company information
- external communication channels

Genuine postings generally provide:

- detailed job descriptions
- company information
- clear responsibilities
- realistic hiring processes

## Model Limitations

Although the model performs well, several limitations remain.

- It relies primarily on textual content.

- Images and company logos are not analyzed.

- Company reputation is not verified.

- External company websites are not validated.

- Predictions may be affected by unusual wording.

## Future Improvements

Potential future enhancements include:

- Transformer-based language models (BERT)

- Explainable AI using SHAP or LIME

- Company reputation verification

- Website trust analysis

- Salary anomaly detection

- Continuous retraining with new job postings
